## 5 Filtering and Boolean Indexing

```python
df[df['age']>28] #simple condition ---->
```

```python
df[(df['age']>25) & (df['city']== 'NY')] #multiple condition ---->combine with & (and)/ |or)
```

```python
df[(df['age']>36) | (df['city']=='NY')]
```

```python
df[(df['name']=='Bob')| (df['city']=='NY')]
```

```python
df[df['city'].isin(['NY','LA'])] #filter rows where a column matches a set of values
```

```python
df.query('age >28 and city =="NY"') #filter using a readable string expression
```

```python
df.where(df['age'] >28) #keep matching values ,replace other with NAN
```

```python
df.mask(df['age'] >28) #opposite of where() ,replace matching values with NAN
```

## 7 Handling Missing Data

```python
df_na =pd.DataFrame({'a':[1,np.nan,3],'b':[np.nan,2,3]})
df_na
```

```python
#  .isna() ---> Boolean mask of missing values
df_na.isna()
```

```python
# .notna() ----> opposite of .isna()
df_na.notna()
```

```python
#df_na.isnull().sum() ---> Count of missing values per columns
df_na.isnull().sum()
```

```python
# .dropna()-----> Drop rows containing any NaN
df_na.dropna()
```

```python
# .dropna(axis=1)-----> Drop columns containing any NaN
df_na.dropna(axis=1)
```

```python
#  .fillna() ---->  fill missing values with constant
df_na.fillna(0)
```

```python
#   .ffill() -----> Forward fill missing values
df_na.ffill()
```

```python
#   .bfill() -----> Backward fill missing values
df_na.bfill()
```

# 8 Sorting And Ranking

```python
df=pd.DataFrame({
    'name':['Alice','Bob','Charlie','David'],
    'age':[25,30,35,40],
    'city':['NY','LA','SF','NY']
})
df
```

```python
# .sort_values() ----> Arrange the values in ascending order
df.sort_values('age')
```

```python
df.sort_values(['city','age'],ascending=[True,False])
```

```python
# .sort_index() ----> Sort rows by the index
df.sort_index()
```

```python
# .rank() ----> Rank values within a column
df['age'].rank
```

```python
# .nlargest() ----> Top n rows by a column
df.nlargest(2,'age')
```

```python
# .nsmallest()----> Bottom n rows by a column
df.nsmallest(3,'age')
```

# 9 Aggregation and Statistics

```python
# .sum() / .mean() / .median() ------> Basic Aggregations
df['age'].sum(), df['age'].mean(), df['age'].median()
```

```python
# .std() / .var() ----> Spread of data
df['age'].std(), df['age'].var()
```

```python
# .min() / .max() ----> Smallest and Largest value
df['age'].min(), df['age'].max()
```

```python
# .count() ---> Number of not null values
df['age'].count()
```

```python
# .agg()----> Apply a multiple aggregation functions at once
df.agg({'age':['mean','max','min']})
```

```python
# .corr() ----->correlation matrix of numeric columns
df.corr(numeric_only=True)
```

```python
# .mode() ---->Most frequently occuring value
df['city'].mode()
```

## 10 GroupBy Operation

```python
g=df.groupby('city')
```

```python
#.size() ----->number of rows in each group
```

```python
g.size()
```

```python
#Group then aggregate a single column
```

```python
g['age'].mean()
```

```python
#.agg() on a groupby-----> multiple aggregation per group
```

```python
g.agg({'age':['mean','sum']})
```

```python
#.apply() on a groupby-----> custom function per group
```

```python
g.apply(lambda x:x ['age'].max()-x['age'].min())
```

```python
add = lambda a,b : a+b
```

```python
add(4,5)
```

```python
#iterating over group with a for loop
```
